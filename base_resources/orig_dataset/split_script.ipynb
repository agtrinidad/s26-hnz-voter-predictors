```python
import pandas as pd 
import string
```

```python
orig_df = pd.read_csv('CCES24_Common_OUTPUT_vv_topost_final.csv')
```

```python
orig_df.shape
```

```python
split_part = int(8)
divider = int(orig_df.shape[0]/split_part)
orig_dict = {}
letter_list = string.ascii_lowercase

for i in range(split_part):
    if i == 0:
        orig_dict[f"orig_df_{i}"] = orig_df[:divider]
    else:
        orig_dict[f"orig_df_{i}"] = orig_df[(divider*i):(divider*(i+1))]
    orig_dict[f"orig_df_{i}"].to_csv(f'CCES24_original_{letter_list[i]}.csv')
    
```
